```python
for col in df: print(col, "|", df[col].dtype, "|", df[col].unique()[:5])
```

```python
df["Sex"] = df["Sex"].map({"M": 1, "F": 0})
df = pd.get_dummies(df, columns=["ChestPainType"], drop_first=True, dtype=int)
df["RestingECG"] = df["RestingECG"].map({"Normal": 1, "ST": 0, "LVH": 2})
df["ST_Slope"] = df["ST_Slope"].map({"Up": 0, "Flat": 1, "Down": 2})
df["ExerciseAngina"] = df["ExerciseAngina"].map({"Y": 1, "N": 0})
df
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop(columns=["HeartDisease"])
y=df["HeartDisease"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
x_train
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score= accuracy_score(y_test,y_pred)
accuracy_score
```

```python
classification_report=classification_report(y_test,y_pred)
classification_report
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logic_model=LogisticRegression(random_state=42)
logic_model.fit(x_train,y_train)
```

```python
logic_ypred=logic_model.predict(x_test)
logic_ypred
```

```python
from sklearn.metrics import accuracy_score

a = accuracy_score(y_test, logic_ypred)
print("LOGISTIC REGRESSION Accuracy Score:", a)
```
